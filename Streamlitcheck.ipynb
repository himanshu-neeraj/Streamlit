```python
import streamlit as st

st.set_page_config(page_title="Student Information System", page_icon="🎓")

st.title("🎓 Student Information System")

st.write(
    "Enter student details below and click the submit button to view the information."
)

student_name = st.text_input("Enter Student Name")

roll_number = st.number_input(
    "Enter Roll Number",
    min_value=1,
    step=1
)

course = st.selectbox(
    "Select Course",
    ["B.Sc AI", "BCA", "B.Tech", "B.Com", "MBA", "Other"]
)

subjects = st.multiselect(
    "Select Subjects",
    [
        "Python",
        "Java",
        "DBMS",
        "Machine Learning",
        "Cloud Computing",
        "Data Science"
    ]
)

marks = st.slider(
    "Enter Marks",
    0,
    100,
    50
)

if st.button("Submit"):

    if student_name.strip() == "":
        st.error("Please enter student name")

    elif len(subjects) == 0:
        st.error("Please select at least one subject")

    else:
        st.success("Student details submitted successfully!")

        st.subheader("Student Details")

        st.write("**Student Name:**", student_name)
        st.write("**Roll Number:**", roll_number)
        st.write("**Course:**", course)
        st.write("**Subjects:**", ", ".join(subjects))
        st.write("**Marks:**", marks)
```
